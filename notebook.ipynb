## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana tren jumlah penyewaan sepeda setiap bulan selama tahun 2011–2012, dan bagaimana perubahannya dari tahun 2011 ke tahun 2012?
- **Pertanyaan 2:** Bagaimana perbandingan jumlah penyewaan sepeda oleh pengguna kasual (casual) dan pengguna terdaftar (registered) selama tahun 2011–2012, serta kelompok pengguna mana yang memberikan kontribusi terbesar terhadap total penyewaan?

## Data Wrangling

### Gathering Data

```python
# Memuat dataset Bike Sharing
day_df = pd.read_csv('data/day.csv')

# Menampilkan lima baris pertama
day_df.head()
```

#### **Insight Gathering Data:**

### Assessing Data

#### 1. Informasi dan Tipe Data Dataset

```python
print("Informasi Dataset:")
print(f"Jumlah baris: {day_df.shape[0]}")
print(f"Jumlah kolom: {day_df.shape[1]}")
print("\nDetail tipe data setiap kolom:")
day_df.info()
```

#### 2. Pemeriksaan Missing Value

```python
print("Pemeriksaan Missing Value:")

missing_data = day_df.isna().sum()

missing_report = pd.DataFrame({
    "Jumlah Missing Value": missing_data,
    "Persentase (%)": (missing_data / len(day_df) * 100).round(2)
})

print(missing_report.to_string())
print(f"\nTotal missing value: {missing_data.sum()}")
```

#### 3. Pemeriksaan Data Duplikat

```python
print("Pemeriksaan Data Duplikat:")

duplicate_rows = day_df.duplicated().sum()
duplicate_dates = day_df["dteday"].duplicated().sum()

print(f"Jumlah baris duplikat: {duplicate_rows}")
print(f"Jumlah tanggal duplikat: {duplicate_dates}")
```

#### 4. Pemeriksaan Outlier Data

```python
print("Pemeriksaan Outlier pada Jumlah Penyewaan:")

Q1 = day_df["cnt"].quantile(0.25)
Q3 = day_df["cnt"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = day_df[
    (day_df["cnt"] < lower_bound) |
    (day_df["cnt"] > upper_bound)
]

print(f"Kuartil pertama (Q1): {Q1:.2f}")
print(f"Kuartil ketiga (Q3): {Q3:.2f}")
print(f"Nilai IQR: {IQR:.2f}")
print(f"Batas bawah: {lower_bound:.2f}")
print(f"Batas atas: {upper_bound:.2f}")
print(f"Jumlah outlier: {len(outliers)}")
```

#### 5. Pemeriksaan Inaccurate Value

```python
print("Pemeriksaan Inaccurate Value:")

# Memeriksa nilai negatif pada jumlah penyewaan
count_columns = ["casual", "registered", "cnt"]

negative_counts = (day_df[count_columns] < 0).sum()

# Memeriksa nilai di luar rentang normalisasi 0–1
normalized_columns = ["temp", "atemp", "hum", "windspeed"]

out_of_range = (
    (day_df[normalized_columns] < 0) |
    (day_df[normalized_columns] > 1)
).sum()

print("\nNilai negatif pada kolom penyewaan:")
print(negative_counts.to_string())

print("\nNilai di luar rentang 0–1:")
print(out_of_range.to_string())

print(
    "\nTotal nilai yang terindikasi tidak wajar:",
    int(negative_counts.sum() + out_of_range.sum())
)
```

#### 6. pemeriksaan Inconsistent Value

```python
print("Pemeriksaan Inconsistent Value:")

# Memeriksa kesesuaian total penyewaan
day_df["calculated_cnt"] = (
    day_df["casual"] + day_df["registered"]
)

inconsistent_data = day_df[
    day_df["cnt"] != day_df["calculated_cnt"]
]

print("Jumlah data tidak konsisten:", len(inconsistent_data))

if not inconsistent_data.empty:
    print("\nData yang tidak konsisten:")
    print(
        inconsistent_data[
            ["dteday", "casual", "registered", "cnt", "calculated_cnt"]
        ].to_string(index=False)
    )
else:
    print("Seluruh total penyewaan sesuai dengan penjumlahan pengguna.")
    
# Menghapus kolom bantu agar dataset tetap seperti semula
day_df.drop(columns=["calculated_cnt"], inplace=True)
```

#### **Insight Assessing Data:**
- Dataset memiliki 731 baris dan 16 kolom. Kolom `dteday` masih bertipe `object` dan perlu diubah menjadi `datetime`.
- Tidak ditemukan missing value maupun data duplikat.
- Pemeriksaan IQR tidak menemukan outlier pada kolom `cnt`.
- Tidak ditemukan nilai penyewaan negatif atau nilai normalisasi di luar rentang 0–1.
- Nilai `cnt` konsisten dengan penjumlahan `casual` dan `registered`

### Cleaning Data

#### 1. Mengubah Tipe Data kolom Tanggal 

```python
# Mengubah tipe data kolom tanggal
day_df["dteday"] = pd.to_datetime(day_df["dteday"])

# Memeriksa tipe data setelah cleaning
print("Tipe data dteday setelah cleaning:", day_df["dteday"].dtype)
```

#### 2. Mapping Data menjadi Kategorikal 

```python
# Mengubah kode numerik menjadi kategori
day_df["tahun"] = day_df["yr"].map({
    0: "2011",
    1: "2012"
})

day_df["bulan"] = day_df["mnth"].map({
    1: "Januari",
    2: "Februari",
    3: "Maret",
    4: "April",
    5: "Mei",
    6: "Juni",
    7: "Juli",
    8: "Agustus",
    9: "September",
    10: "Oktober",
    11: "November",
    12: "Desember"
})

day_df["musim"] = day_df["season"].map({
    1: "Musim Semi",
    2: "Musim Panas",
    3: "Musim Gugur",
    4: "Musim Dingin"
})

day_df["kondisi_cuaca"] = day_df["weathersit"].map({
    1: "Cerah / Sedikit Berawan",
    2: "Berkabut / Berawan",
    3: "Hujan atau Salju Ringan",
    4: "Hujan Lebat / Salju / Kabut"
})

day_df["jenis_hari"] = day_df["workingday"].map({
    0: "Hari Nonkerja",
    1: "Hari Kerja"
})

# Memeriksa hasil mapping
day_df[[
    "yr", "tahun",
    "mnth", "bulan",
    "season", "musim",
    "weathersit", "kondisi_cuaca",
    "workingday", "jenis_hari"
]].head()
```

#### 3. Memperbaiki Nama Kolom  

```python
# Memperbaiki nama kolom agar lebih mudah dipahami
day_df = day_df.rename(columns={
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "weathersit": "weather_situation",
    "atemp": "feels_like_temp",
    "hum": "humidity",
    "cnt": "total_rentals"
})

# Memeriksa nama kolom setelah diperbaiki
print("Nama kolom setelah cleaning:")
print(day_df.columns.tolist())
```

#### **Insight Cleaning Data:**
- Kolom `dteday` telah diubah menjadi `date` dengan tipe data `datetime64[ns]` agar dapat digunakan dalam analisis waktu.
- Nama beberapa kolom telah diperjelas, seperti `yr` menjadi `year`, `mnth` menjadi `month`, dan `cnt` menjadi `total_rentals`.
- Kode numerik pada kolom tahun, bulan, musim, dan cuaca telah dipetakan menjadi label yang lebih mudah dipahami untuk analisis selanjutnya.
- Tidak dilakukan pembersihan tambahan karena hasil pemeriksaan sebelumnya tidak menunjukkan masalah kualitas data lain yang memerlukan perbaikan.

## Exploratory Data Analysis (EDA)

### Eksplorasi Tren Penyewaan, Karakteristik Pengguna dan  Penyewaan  Berdasarkan Kondisi Cuaca

#### 1. Analisis Tren Penyewaan Sepeda Bulanan

```python
# Menghitung total penyewaan sepeda setiap bulan berdasarkan tahun
monthly_rentals = (
    day_df.groupby(["year", "month"])["total_rentals"]
    .sum()
    .reset_index()
)

# Mengurutkan data berdasarkan tahun dan bulan
monthly_rentals = monthly_rentals.sort_values(
    ["year", "month"]
)

# Menampilkan hasil analisis
print("Total Penyewaan Sepeda per Bulan:")
print(
    monthly_rentals.to_string(index=False)
)
```

#### 2. Analisis Perbandingan Pengguna Casual dan Registered

```python
# Menghitung total penyewaan berdasarkan jenis pengguna
user_comparison = pd.DataFrame({
    "Jenis Pengguna": ["Casual", "Registered"],
    "Total Penyewaan": [
        day_df["casual"].sum(),
        day_df["registered"].sum()
    ]
})

# Menghitung persentase penyewaan setiap jenis pengguna
user_comparison["Persentase (%)"] = (
    user_comparison["Total Penyewaan"]
    / user_comparison["Total Penyewaan"].sum()
    * 100
).round(2)

# Menampilkan hasil analisis
print("Perbandingan Total Penyewaan Berdasarkan Jenis Pengguna:")
print(user_comparison.to_string(index=False))
```

#### 3. Analisis Penyewaan Sepeda Berdasarkan Kondisi Cuaca

```python
# Menganalisis penyewaan berdasarkan kondisi cuaca
weather_order = [
    "Cerah / Sedikit Berawan",
    "Berkabut / Berawan",
    "Hujan atau Salju Ringan",
    "Hujan Lebat / Salju / Kabut"
]

weather_comparison = (
    day_df.groupby("kondisi_cuaca")["total_rentals"]
    .agg(
        Jumlah_Hari="count",
        Total_Penyewaan="sum",
        Rata_rata_Penyewaan="mean"
    )
    .reindex(weather_order)
    .rename_axis("Kondisi Cuaca")
    .reset_index()
)

# Mengatur nama kolom
weather_comparison = weather_comparison.rename(columns={
    "Jumlah_Hari": "Jumlah Hari",
    "Total_Penyewaan": "Total Penyewaan",
    "Rata_rata_Penyewaan": "Rata-rata Penyewaan"
})

# Mengisi kategori yang tidak ditemukan dengan nol
weather_comparison[[
    "Jumlah Hari",
    "Total Penyewaan",
    "Rata-rata Penyewaan"
]] = weather_comparison[[
    "Jumlah Hari",
    "Total Penyewaan",
    "Rata-rata Penyewaan"
]].fillna(0)

weather_comparison["Rata-rata Penyewaan"] = (
    weather_comparison["Rata-rata Penyewaan"].round(2)
)

print("Perbandingan Penyewaan Sepeda Berdasarkan Kondisi Cuaca:")
print(weather_comparison.to_string(index=False))
```

#### **Insight EDA:**
- Data penyewaan dikelompokkan berdasarkan tahun dan bulan untuk mengetahui pola penyewaan sepeda selama periode 2011–2012.
- Total penyewaan pengguna casual dan registered dihitung untuk mengetahui kontribusi masing-masing jenis pengguna terhadap keseluruhan penyewaan.
- Data penyewaan dikelompokkan berdasarkan kondisi cuaca untuk membandingkan jumlah hari, total penyewaan, dan rata-rata penyewaan pada setiap kategori cuaca.

## Analisis Lanjutan

### Analisis Lanjutan: Binning dan Manual Grouping

Analisis ini dilakukan untuk mengelompokkan jumlah penyewaan sepeda harian menjadi beberapa tingkat permintaan, kemudian membandingkannya berdasarkan kondisi cuaca dan musim.

#### 1. Menentukan kategori permintaan

```python
# Mengelompokkan tingkat permintaan penyewaan sepeda
day_df["kategori_permintaan"] = pd.qcut(
    day_df["total_rentals"],
    q=3,
    labels=["Rendah", "Sedang", "Tinggi"]
)

# Menampilkan beberapa hasil pengelompokan
print("Contoh Kategori Permintaan Penyewaan:")
print(
    day_df[["date", "total_rentals", "kategori_permintaan"]]
    .head(10)
    .to_string(index=False)
)
```

#### 2. Tingkat Permintaan Berdasarkan Kondisi Cuaca

```python
# Membandingkan tingkat permintaan berdasarkan kondisi cuaca
weather_demand = (
    day_df.groupby(
        ["kondisi_cuaca", "kategori_permintaan"],
        observed=False
    )
    .size()
    .unstack(fill_value=0)
    .reindex(
        columns=["Rendah", "Sedang", "Tinggi"],
        fill_value=0
    )
)

# Mengatur urutan kategori kondisi cuaca
weather_demand = weather_demand.reindex([
    "Cerah / Sedikit Berawan",
    "Berkabut / Berawan",
    "Hujan atau Salju Ringan",
    "Hujan Lebat / Salju / Kabut"
], fill_value=0)

print("Tingkat Permintaan Berdasarkan Kondisi Cuaca:")
print(weather_demand.to_string())
```

#### **Insight Analisis Lanjutan:**

- Pada kondisi cerah atau sedikit berawan, terdapat 182 hari dengan tingkat permintaan tinggi, lebih banyak dibandingkan kondisi cuaca lainnya.
- Pada cuaca berkabut atau berawan, kategori permintaan rendah paling banyak ditemukan, yaitu 104 hari. Sementara itu, pada kondisi hujan atau salju ringan, terdapat 19 hari dengan permintaan rendah dan 2 hari dengan permintaan sedang, tanpa adanya permintaan tinggi.
- Hasil pengelompokan menunjukkan bahwa permintaan tinggi lebih sering ditemukan pada kondisi cuaca cerah, sedangkan permintaan rendah lebih banyak ditemukan saat cuaca kurang mendukung. Namun, hasil ini menunjukkan pola hubungan dan tidak membuktikan bahwa cuaca menjadi satu-satunya faktor yang memengaruhi penyewaan.
